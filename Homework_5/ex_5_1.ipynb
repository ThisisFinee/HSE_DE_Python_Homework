```python
from account import Account
import time
```

```python
acc = Account("Иван Rest", 1000)
time.sleep(1)
acc.deposit(500) # Пополнение
time.sleep(1)
acc.withdraw(200) # Удачное снятие
time.sleep(1)
acc.withdraw(2000) # Неудачное снятие

print(f"Текущий баланс: {acc.get_balance()}")
print("\nИстория операций:")
for op in acc.get_history():
    print(op)

acc.plot_history()
```

```python
try:
    acc.deposit(-300) # Ошибка при пополнении
except ValueError as e:
    print(e)
try:
    acc.withdraw(-300) # Ошибка при снятии
except ValueError as e:
    print(e)
```
